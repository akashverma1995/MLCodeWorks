```python
# VECTOR OPERATIONS
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("a + b =", a + b)               # Element-wise addition
print("a - b =", a - b)               # Element-wise subtraction
print("a * b =", a * b)               # Element-wise multiplication
print("a / b =", a / b)               # Element-wise division
print("Dot product a . b =", np.dot(a, b))  # Dot product
print("Cross product a x b =", np.cross(a, b))  # Cross product
print()
```

```python
# TENSOR OPERATIONS
ta = torch.tensor([1, 2, 3])
tb = torch.tensor([4, 5, 6])

print("ta + tb =", ta + tb)               # Element-wise addition
print("ta - tb =", ta - tb)               # Element-wise subtraction
print("ta * tb =", ta * tb)               # Element-wise multiplication
print("ta / tb =", ta / tb)               # Element-wise division
print("Dot product ta . tb =", torch.dot(ta, tb))  # Dot product
print("Cross product ta x tb =", torch.cross(ta, tb))  # Cross product

```

```python
# VECTOR MAGNITUDE AND DIRECTION
v = torch.tensor([3, 4], dtype=torch.float32)
magnitude = torch.norm(v)
print(magnitude)   # 5.0
angle_rad = torch.atan2(v[1], v[0])
angle_deg = np.degrees(angle_rad.item())
print(angle_deg)   # ~53.13 degrees
unit_vector = v / torch.norm(v)
print(unit_vector)
print(torch.norm(unit_vector))  # 1.0
```

# Partial derivates and Gradents

## Partial Derivatives — Local Thinking

**Intuition:**  
A partial derivative answers the question:

> *“If I change one variable by a tiny amount while keeping all other variables fixed, how does the output change?”*

### Visual Intuition: A Mountain Surface

Imagine a mountain represented by a surface:

- **x** → movement east–west  
- **y** → movement north–south  
- **z** → height (elevation)

Taking a **partial derivative** means measuring the slope of the mountain:
- Along the **x-direction** (holding *y* constant), or  
- Along the **y-direction** (holding *x* constant)

You are not moving freely on the surface —  
you are slicing it in one direction and observing the slope locally.


## Visual Thinking

### Partial Derivative vs Gradient

- **Partial derivative** → slope along **one axis**
- **Gradient** → an **arrow on the surface** showing direction

### Mountain + Compass Analogy

- **Partial derivative** = checking the slope along a single path  
- **Gradient** = a **GPS/compass** telling you the *best direction to move*

You don’t guess which way is down — the gradient tells you.

---

## Why Gradients Matter in Machine Learning

### Loss Function

\[
L(w_1, w_2, \dots, w_n)
\]

- Each \( w_i \) is a **model weight**
- All weights together form a point in a **high-dimensional space**

### Core Idea

The **gradient of the loss** answers:

> *“In which direction should **all weights move together** to reduce the loss as fast as possible?”*

That’s why training works:
- Gradient → direction  
- Learning rate → step size  
- Repeated steps → model learns


```python
w = 10.0
lr = 0.1

for _ in range(20):
    grad = 2*w        # derivative of w^2
    w -= lr * grad

print(w)
```
