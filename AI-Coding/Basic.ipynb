### Vector

Direction works only for 2d or 3d vector


```python
v = np.array([3, 4])
magnitude = np.linalg.norm(v)
print(magnitude)   # 5.0

angle_rad = np.arctan2(4, 3)
angle_deg = np.degrees(angle_rad)

print(angle_deg)   # ~53.13 degrees

unit_vector = v / np.linalg.norm(v)
print(unit_vector)
print(np.linalg.norm(unit_vector))  # 1.0# 2-D ARRAY or MATRIX
```

```python
# VECTOR OPERATIONS
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("a + b =", a + b)               # Element-wise addition
print("a - b =", a - b)               # Element-wise subtraction
print("a * b =", a * b)               # Element-wise multiplication
print("a / b =", a / b)               # Element-wise division
print("Dot product a . b =", np.dot(a, b))  # Dot product
print("Cross product a x b =", np.cross(a, b))  # Cross product
print()
```

```python
# TENSOR OPERATIONS
ta = torch.tensor([1, 2, 3])
tb = torch.tensor([4, 5, 6])

print("ta + tb =", ta + tb)               # Element-wise addition
print("ta - tb =", ta - tb)               # Element-wise subtraction
print("ta * tb =", ta * tb)               # Element-wise multiplication
print("ta / tb =", ta / tb)               # Element-wise division
print("Dot product ta . tb =", torch.dot(ta, tb))  # Dot product
print("Cross product ta x tb =", torch.cross(ta, tb))  # Cross product

```

```python
# VECTOR MAGNITUDE AND DIRECTION
v = torch.tensor([3, 4], dtype=torch.float32)
magnitude = torch.norm(v)
print(magnitude)   # 5.0
angle_rad = torch.atan2(v[1], v[0])
angle_deg = np.degrees(angle_rad.item())
print(angle_deg)   # ~53.13 degrees
unit_vector = v / torch.norm(v)
print(unit_vector)
print(torch.norm(unit_vector))  # 1.0
```
